# Manage a To Do List

```python
to_do_list= ["wake up", "drink water", "eat breakfast", "study"]

## Add task
to_do_list.append("take a shower")
to_do_list.append("take lunch")

## Remove task
to_do_list.remove("eat breakfast")

## checking if a task is in the list

if "drink water" in to_do_list:
    print("Dont forget to drink water!")

print("To Do List remaining")
for task in to_do_list:
    print(f"-{task}")




```

# Organizing Student Grades

```python
grades= [10,5,6,3,10,8,9,1]

## Adding a new grade
grades.append(7)

## Calculating the average grade
average_grade= sum(grades)/len(grades)
print(f"The average grade is: {average_grade:.2f}")

## Finding the highest and lowest grade
highest_grade= max(grades)
lowest_grade= min(grades)
print(f"The highest grade is: {highest_grade}")
print(f"The lowest grade is: {lowest_grade}")
```

# Managing An Inventory

```python
inventory=["pen", "pencil", "notebook", "eraser", "ruler"]

## Adding a new item
inventory.append("marker")

## Removing an item
inventory.remove("ruler")

## Checking if an item is in stock
item= "notebook"
if item in inventory:
    print(f"{item} is in stock.")
else:
    print(f"{item} is out of stock.")


## Printing the inventory
print("Inventory List:")
for item in inventory:
    print(f"- {item}")
```

# Collecting User Feedback

```python
feedback= ["Great service!", "Very satisfied", "Could be better", "Excellent experience"]

## Adding new feedback
feedback.append("Not happy with the service.")

## Counting specific feedback
positive_feedback_count= sum(1 for comment in feedback if "great" in comment.lower()or "excellent" in comment.lower())
print(f"Positive feedback Count is: {positive_feedback_count}")

##Printing all feedbacks
print("User Feedbacks:")
for comment in feedback:
    print(f"- {comment}")
```
