#### 2. Store data in key-value pairs.

#### 4. Values can be of any type.

## # Accessing Dictionary Elements

```python
student={"name":"debangshi",
         "age": 23,
         "course": "Data Analytics"}

print(student['name'])
print(student["age"])


## Accessing using get() method
print(student.get("name"))
print(student.get("course"))


print(student.get("phone"))     ## It will return 'None' if the key is not found in the dictionary.
print(student.get("phone", "Not Found"))   ## It will return "Not FOund" if the key is not found in the dictionary.
```

### Modifying Dictionary Elements
### Dictionary are mutable, so you can add, update or delete elements

```python
student={"name":"debangshi",
         "age": 23,
         "course": "Data Analytics"}

print(student)

student["age"]=24    ## Updating the value of the key "age".
print(student)

student["phone"]= 7384563401    ## Adding a new key-value pair to the dictionary.
print(student)

del student["age"]  ## Deleting a key-value pair from the dictionary using del keyword.
print(student)
```

### Dictionary methods

```python
keys=student.keys()   ## get all the keys of the dictionary using keys() method.
print(keys)

values=student.values()  ## get all the values of the dictionary using values() method.
print(values)

items=student.items()  ## get all the key-value pairs of the dictionary using items() method.
print(items)
```

### # Shallow copy

##### Copying a dictionary using assignment operator will not create a new copy of the dictionary. It will create a reference to the original dictionary. So, if we change the value of a key in the copied dictionary, it will also change the value of the key in the original dictionary.

```python
student={"name":"debangshi",
         "age": 23,
         "course": "Data Analytics"}

student_copy= student
print(student)
print(student_copy)


##changing the value of the key
student_copy["name"]= "John"
print(student)
print(student_copy)
```

##### But if we want to create a new copy of the dictionary, we can use the copy() method to create a new copy of the dictionary. So if we chanage the value of a key in the copied dictionary, it will not change the value of the key in the original dictionary.

```python
student={"name":"debangshi",
         "age": 23,
         "course": "Data Analytics"}

student_copy1=student.copy()
print(student)
print(student_copy1)

student["name"]="John"
print(student)
print(student_copy1)

```

### Iterating over dictionaries

```python
## 1. Iterating over keys

for keys in student.keys():
    print(keys)
```

```python
## 2. Iterating over values

for values in student.values():
    print(values)
```

```python
## 3. Iterating over key-value pairs

for keys, values in student.items():
    print(f"{keys}: {values}")
```

### Nested Dictionaries

```python
students={
    "student1": {
        "name": "Alice",
        "age": 20,
        "skills": ["Python", "Data Analytics"]
    },
    "student2": {
        "name": "Bob",
        "age": 22,
        "skills": ["SQL", "Power BI"]
    }
}

print(students)
```

#### Access nested dictionaries elements

```python
print(students["student2"]["name"])
print(students["student1"]["skills"])
```

```python
print(students.items())
```

### Iterating over nested dictionaries

```python
for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key,value in student_info.items():
        print(f"{key}: {value}")
```

## Dictionary Comprehension

```python
squares= {x: x**2 for x in range (5)}
print(squares)
```

```python
## Condition dictionary comprehension

even_squares= {x: x**2 for x in range(11) if x%2==0}
print(even_squares)
```

### Q. Use a dictionary to count the frequency of elements in list

```python
lst= [1,2,2,2,3,5,4,6,6,9,2,4,7,3,5,10]

unique_nos= list(set(lst))
print(unique_nos)
freq={}

for i in range(len(unique_nos)):
    c=0
    for j in range(len(lst)):
        if unique_nos[i] == lst[j]:
            c=c+1
    freq[unique_nos[i]]=c      ## Adding the unique number as key and its frequency as value in the dictionary.
print(freq)

```

### Q. Merge 2 dictionaries into one

```python
dict1= {1:"open",
        2:"two",
        3:"three"}

dict2= {0:"false",
        1:"True"}

## using keyword arguments
## Keyword arguments are used to pass the values to the function. The keyword arguments are passed in the form of key-value pairs. 

merged_dict={**dict1, **dict2}    ## here we are using the unpacking operator ** to unpack the key-value pairs of the dictionaries dict1 and dict2 and merge them into a new dictionary merged_dict.
print(merged_dict)
```
