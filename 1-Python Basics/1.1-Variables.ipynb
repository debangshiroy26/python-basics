### Naming Conventions


```python
## Variable names should be descriptive.
##They must start with a letter or an underscore (_). And can contain letters, numbers and underscores.
## variable names are case sensitive.

first_name="Debangshi"
last_name="Roy"

## Invalid variable names

#2age=30
#first-name="Debangshi"
#@name="Deb"
```

```python
## Python is a dynamically typed lang
## Type of a variable is determined at runtime and can change as the program executes.

age= 23 #int
height= 5.3 #float
name="Debangshi" #string
is_student=True #boolean
```

```python
## Type Conversion

age= 23
print(type(age))

age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
age_str='23'
age= int(age_str)
print(type(age))

#name="Deb"
#int(name)  #this will give an error as we cant convert string to int if the string is not a number.

height=5.3
print(type(height))
height_int=int(height)
print(type(height_int))
```

```python
## Dynamic typing 
## Puthon allows the type of a variable to change as the program executes.

var=10  #int
print(var, type(var))

var="Hello"  #string
print(var, type(var))

var=5.5  #float
print(var, type(var))
```

```python
## input
## input() func. always returns a string.
## if u want to take an integer input, u need to convert the string to int using int()func.

age=int(input("What is the age?"))
print(age,type(age))
```

### SIMPLE CALCULATOR

```python
num1=float(input("Enter first number: "))
num2=float(input("Enter second number:"))

sum= num1+num2
difference= num1-num2
product=num1*num2
quotient=num1/num2

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
