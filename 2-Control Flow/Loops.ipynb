### For Loop

```python
range(5)

## range is a built-in function in python that generates a sequence of numbers.
## the number itself is not included in the range.
```

```python
## for Loop

for i in range(5):
    print(i)

```

```python
for i in range(1,6):
    print(i)
```

```python
## range() can take 3 arguments: start, stop, step.
# 1. start: starting number of the sequence.
# 2. stop: generate numbers upto but not including this number.
# 3. step: difference between each number in the sequence.

for i in range(1,20,2):
    print(i)
```

```python
for i in range(20,1,-2):
    print(i)
```

```python
## in strings, we can use for loop to iterate through each character in the string.

str= "Hello World"

for i in str:
    print(i)
```

### While Loop

```python
## while loop continues to execute a block of code as long as the condition is true.

count=0

while count<10:
    print(count)
    count+=1
```

### Loop Control Statements

## 1. Break statement

```python
## it is used to exit the loop permanently when a certain condition is met.

for i in range(10):
    if i==5:
        break
    print(i)
```

## 2. Continue statement

```python
## it is used to skip the current iteration of the loop and move on to the next iteration.

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
for i in range(5):
    if i==3:
        continue
    print(i)
```

## 3. Pass Statement

```python
## it is a null operation, it is used when a statement is required syntactically but you do not want any command or code to execute.
## it is often used as a placeholder for future code.
## it just do nothing and move on to the next iteration of the loop.

for i in range(10):
    if i==5:
        pass
    print(i)
```

### Nested Loops

```python
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i: {i} and j:{j}")

```

## 1. Calculate the sum of first N natural numbers using a while and for loop

```python
## 1. while loop

n= 10
sum=0
count=1

while count<=n:
    sum= sum+count
    count= count+1

print("Sum of first 10 natural numbers is:", sum)

    
```

### 2. Prime Number

```python
num= int(input("Enter a number"))
count=0
for i in range(1,num+1):
    if num%i==0:
        count=count+1
    else:
        pass

if count==2:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```
